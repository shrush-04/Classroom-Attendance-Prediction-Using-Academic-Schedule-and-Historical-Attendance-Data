# Step 1: Classroom Attendance Data Validation
This notebook runs the 16 validation rules on the raw classroom attendance log.

### Validation Rules
1. Required columns presence
2. `Lecture_ID` format and duplicate check
3. ISO Date format check (`YYYY-MM-DD`)
4. Time format check (`HH:MM`)
5. Chronological order check
6. Capacity check (`Students_Present` <= `Total_Enrolled_Students`)
7. Non-negative counts check
8. Percentage calculation validation
9. Missing values check in required columns
10. Duplicates check
11. Categorical levels check
12. Subject name consistency
13. Section consistency
14. Faculty_ID encoding consistency
15. No personal information columns
16. No names, email IDs, or roll numbers in content cells

```python
import os
import sys
import pandas as pd

# Include src directory
sys.path.append('../src')
import validate_raw_data

raw_data_path = '../data/raw/raw_lecture_attendance.csv'

if not os.path.exists(raw_data_path):
    print("=========================================================================")
    print("WARNING: Raw attendance data is not present in data/raw/.")
    print("Please copy academic schedule records into:")
    print(os.path.abspath(raw_data_path))
    print("before executing validation rules.")
    print("=========================================================================")
else:
    success, errors = validate_raw_data.run_validation(raw_data_path, output_dir='../outputs')
    if success:
        print("Validation PASSED successfully. Quality reports written to outputs/.")
    else:
        print(f"Validation FAILED with {len(errors)} errors. Inspect reports for details.")
```
