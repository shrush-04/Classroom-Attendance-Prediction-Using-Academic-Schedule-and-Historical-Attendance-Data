# Step 2: Data Cleaning and Feature Engineering
This notebook cleans the raw attendance logs and constructs predictive features while avoiding data leakage.

### Feature Pipeline
- Day of week integer index
- Day of semester index
- Week number
- Time of day (Morning/Afternoon)
- Days since last holiday
- Pre-exam week binary indicator
- Shifted previous lecture attendance percentage (no leakage)
- Gap hours since previous lecture
- Rolling average of previous 3 lectures (no leakage)
- expanding average per subject (no leakage)
