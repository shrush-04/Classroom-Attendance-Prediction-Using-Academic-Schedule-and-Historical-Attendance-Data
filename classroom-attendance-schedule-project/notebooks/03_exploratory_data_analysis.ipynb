# Step 3: Exploratory Data Analysis (EDA)
This notebook analyzes factors affecting attendance and plots distributions, trends, and correlations.

```python
import os
import pandas as pd
import numpy as np

clean_data_path = '../data/processed/cleaned_lecture_attendance.csv'

if not os.path.exists(clean_data_path):
    print("WARNING: Cleaned attendance data is missing. Run validation and cleaning first.")
else:
    df = pd.read_csv(clean_data_path)
    print(f"Dataset contains {len(df)} rows and {df.shape[1]} columns.")
    print("\nDescriptive Statistics:")
    display(df[['Total_Enrolled_Students', 'Students_Present', 'Attendance_Percentage']].describe())
```

### Try Generating Plots
We will attempt to display charts inside this notebook.

```python
try:
    import matplotlib.pyplot as plt
    import seaborn as sns
    %matplotlib inline
    
    if os.path.exists(clean_data_path):
        df = pd.read_csv(clean_data_path)
        
        # Plot distribution
        plt.figure(figsize=(8, 4))
        sns.histplot(df['Attendance_Percentage'], kde=True, color='purple')
        plt.title('Distribution of Attendance Percentage')
        plt.xlabel('Attendance %')
        plt.ylabel('Count')
        plt.show()
        
        # Subject-wise attendance
        plt.figure(figsize=(10, 4))
        sns.boxplot(data=df, x='Subject', y='Attendance_Percentage', palette='Set2')
        plt.title('Subject-wise Attendance Distribution')
        plt.xticks(rotation=45)
        plt.show()
except ImportError:
    print("Matplotlib/Seaborn not installed. Skipping notebook inline plots.")
```
